import numpy as np
from scipy.stats import norm

# Function to perform a one-sample Z-test
def one_sample_z_test(sample_data, population_mean, population_std):
    # Calculate sample mean and sample size
    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)

    # Z-test calculation
    z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

    # Calculate p-value (two-tailed test)
    p_value = 2 * (1 - norm.cdf(abs(z_score)))  # For two-tailed test

import numpy as np

# Function to perform a two-tailed Z-test
def two_tailed_z_test(sample_data, population_mean, population_std):
    # Calculate sample mean and sample size

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def visualize_errors(pop_mean_null, pop_mean_alt, pop_std, sample_size, alpha=0.05):
    z_critical = norm.ppf(1 - alpha / 2)  # Critical Z-value
    se = pop_std / np.sqrt(sample_size)
    
    critical_value_low = pop_mean_null - z_critical * se
    critical_value_high = pop_mean_null + z_critical * se

    x = np.linspace(pop_mean_null - 4 * se, pop_mean_alt + 4 * se, 1000)

    plt.plot(x, norm.pdf(x, pop_mean_null, se), label="Null Hypothesis (H0)", color="blue")
    plt.plot(x, norm.pdf(x, pop_mean_alt, se), label="Alternative Hypothesis (H1)", color="green")

    plt.fill_between(x, 0, norm.pdf(x, pop_mean_null, se), where=(x < critical_value_low) | (x > critical_value_high),
                     color='red', alpha=0.5, label="Type 1 Error (α)")
    plt.fill_between(x, 0, norm.pdf(x, pop_mean_alt, se), where=(x > critical_value_low) & (x < critical_value_high),
                     color='yellow', alpha=0.5, label="Type 2 Error (β)")

    plt.axvline(critical_value_low, color='black', linestyle='--', label=f"Critical Value (low) = {critical_value_low:.2f}")
    plt.axvline(critical_value_high, color='black', linestyle='--', label=f"Critical Value (high) = {critical_value_high:.2f}")

    plt.title("Type 1 and Type 2 Errors")
    plt.legend()
    plt.grid(True)
    plt.show()

import numpy as np
from scipy import stats

def confidence_interval(data, confidence=0.95):
    # Sample statistics
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)  # Sample standard deviation
    sample_size = len(data)

    # T-critical value for the confidence level
    t_critical = stats.t.ppf((1 + confidence) / 2, df=sample_size - 1)

    # Margin of error

    # Confidence interval
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return lower_bound, upper_bound

def bayesian_inference(prior, likelihood, evidence):
    """
    Calculate the posterior probability using Bayes' Theorem.
    
    :param prior: The prior probability of the hypothesis.
    :param likelihood: The likelihood of observing the evidence given the hypothesis.
    :param evidence: The probability of observing the evidence.
    :return: Posterior probability.

    return posterior

import numpy as np
from scipy.stats import f

def f_test(sample1, sample2):
    """
    Perform an F-test to compare the variances of two samples.
    
    :param sample1: First sample data.
    :param sample2: Second sample data.
    :return: F-statistic and p-value.
    """
    # Calculate variances of both samples
    var1 = np.var(sample1, ddof=1)  # Sample variance (use ddof=1 for unbiased estimate)
    var2 = np.var(sample2, ddof=1)
    
    # Calculate F-statistic
    f_statistic = var1 / var2 if var1 > var2 else var2 / var1  # Ensure F-statistic is >= 1
    
    # Degrees of freedom
    df1 = len(sample1) - 1
    df2 = len(sample2) - 1
    
    # Calculate the p-value for the F-statistic
    p_value = 2 * min(f.cdf(f_statistic, df1, df2), 1 - f.cdf(f_statistic, df1, df2))
    
    return f_statistic, p_value

# Example data: Two random samples

sample2 = np.random.normal(loc=0, scale=3, size=50)  # Sample 2 with variance 9

# Perform the F-test
f_stat, p_value = f_test(sample1, sample2)

print(f"F-statistic: {f_stat:.3f}")
print(f"P-value: {p_value:.3f}")

Function to Check Assumptions for ANOVA

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import shapiro, levene, normaltest

def check_anova_assumptions(groups):
    """
    Check the assumptions for ANOVA: normality, independence, and equal variance.
    
    :param groups: List of groups (arrays) to check.
    :return: None (Prints the results of the assumption checks)
    """
    
    # 1. Check Normality using the Shapiro-Wilk test for each group
    print("\nNormality Test (Shapiro-Wilk):")
    for i, group in enumerate(groups, 1):
        stat, p_value = shapiro(group)
        print(f"Group {i}: Stat={stat:.3f}, p-value={p_value:.3f}")
        if p_value < 0.05:
            print(f"Group {i}: Reject null hypothesis (data is not normally distributed)")
        else:
            print(f"Group {i}: Fail to reject null hypothesis (data is normally distributed)")
    
    # 2. Check Equal Variance using Levene's test
    print("\nEqual Variance Test (Levene's test):")
    stat, p_value = levene(*groups)
    print(f"Levene's test statistic: {stat:.3f}, p-value: {p_value:.3f}")
    if p_value < 0.05:
        print("Reject null hypothesis (variances are significantly different)")
    else:
        print("Fail to reject null hypothesis (variances are equal)")

    # 3. Plot Q-Q Plots for visual check of normality
    plt.figure(figsize=(12, 6))
    for i, group in enumerate(groups, 1):
        plt.subplot(1, len(groups), i)
        plt.title(f"Group {i} - Q-Q Plot")
        plt.scatter(np.sort(group), np.random.normal(size=len(group)), alpha=0.5)
    plt.show()

import numpy as np
import scipy.stats as stats

def z_test_proportions(success_a, size_a, success_b, size_b):
    # Sample proportions
    p1 = success_a / size_a
    p2 = success_b / size_b
    
    # Pooled proportion
    p_pooled = (success_a + success_b) / (size_a + size_b)
    
    # Z-test statistic
    z_stat = (p1 - p2) / np.sqrt(p_pooled * (1 - p_pooled) * (1 / size_a + 1 / size_b))
    
    # Calculate two-tailed p-value
    p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))
    
    return z_stat, p_value

# Example data: Two groups
# Group A: 50 successes out of 200
# Group B: 30 successes out of 180
success_a, size_a = 50, 200
success_b, size_b = 30, 180

# Perform Z-test for proportions

# Output results

# Decision

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Function to perform F-test for equal variances
def f_test(data1, data2):
    # Calculate the variances of both datasets
    var1 = np.var(data1, ddof=1)
    var2 = np.var(data2, ddof=1)
    
    # Calculate the F-statistic
    F_statistic = var1 / var2 if var1 > var2 else var2 / var1
    
    # Degrees of freedom
    df1 = len(data1) - 1
    df2 = len(data2) - 1
    
    # Calculate p-value (two-tailed test)
    p_value = 2 * min(stats.f.cdf(F_statistic, df1, df2), 1 - stats.f.cdf(F_statistic, df1, df2))
    
    return F_statistic, p_value, df1, df2

# Simulate two datasets with different variances

data2 = np.random.normal(loc=50, scale=12, size=100)  # Mean = 50, Std Dev = 12

# Perform F-test

print(f"P-value: {p_value:.4f}")
